```python
# country 테이블에서
# GovernmentForm 필드 기준으로 내림차순으로 정렬한 후
# SurfaceArea 필드 기준으로 오름차순 정렬하여 조회
```

```python
conn = pymysql.connect(
    host = 'localhost',
    port = 3306,
    user = 'root',
    password = '0000',
    db = 'world',
    charset = 'utf8mb4'
)
cur = conn.cursor()
query = """
SELECT Name,SurfaceArea,GovernmentForm
FROM country
ORDER BY GovernmentForm DESC, SurfaceArea ASC
"""
cur.execute(query)

result = cur.fetchall()
columns = [col[0] for col in cur.description] 
df = pd.DataFrame(result,columns = columns) 
df
```

```python
# country 테이블에서
# Continent 필드가 'Africa'인 데이터 중에
# Population 기준으로 오름차순 정렬한 후
# Name, Population 데이터
```

```python
query = """
SELECT Name, Population
FROM country
WHERE Continent = 'Africa'
ORDER BY Population ASC
"""
cur.execute(query)
result = cur.fetchall()
columns = [col[0] for col in cur.description] 
df = pd.DataFrame(result,columns = columns) 
df
```

```python
# country 테이블에서 Continent 필드로 그룹화하고
# GNP 평균값을 소수점 2자로 반올림하여 조회하고 이름을 avg_GNP로 출력
```

```python
query = """
SELECT Continent, ROUND(AVG(GNP), 2) AS avg_GNP
FROM country
GROUP BY Continent
"""
cur.execute(query)
result = cur.fetchall()
columns = [col[0] for col in cur.description] 
df = pd.DataFrame(result,columns = columns) 
df
```

```python
# country 테이블에서 Region 필드로 그룹화하고
# 나라의 개수가 15이상 20이하인 데이터를 count_reg로 하여 내림차순으로 조회
```

```python
query = """
SELECT Region, COUNT(*) AS count_reg
FROM country
GROUP BY Region
HAVING COUNT(*) BETWEEN 15 AND 20
ORDER BY count_reg DESC
"""
cur.execute(query)
result = cur.fetchall()
columns = [col[0] for col in cur.description] 
df = pd.DataFrame(result,columns = columns) 
df
```

```python
# country 테이블의 각 국가마다
# 해당 국가에 소속된 도시들의 평균 인구를 city 테이블을 이용하여 조회
```

```python
query = """
SELECT c.Name AS Country, AVG(ci.Population) AS Average_City_Population
FROM country c
JOIN city ci ON c.Code = ci.CountryCode
GROUP BY c.Name
"""
cur.execute(query)
result = cur.fetchall()
columns = [col[0] for col in cur.description] 
df = pd.DataFrame(result,columns = columns) 
df
```

```python
# country 테이블의 Frace의 국가코드를 이용하여
# city 테이블의 'France'의 모든 도시의 이름 출력
```

```python
query = """
SELECT ci.Name AS CityName
FROM city ci
JOIN country c ON ci.CountryCode = c.Code
WHERE c.Name = 'France'
"""
cur.execute(query)
result = cur.fetchall()
columns = [col[0] for col in cur.description] 
df = pd.DataFrame(result,columns = columns) 
df
```

```python
# world DB의 city와 country 테이블과 이용하여
# 도시 인구 비율 계산 함수 (도시인구/해당 국가의 인구)
```

```python
conn = pymysql.connect(
    host = 'localhost',
    port = 3306,
    user = 'root',
    password = '0000',
    db = 'world',
    charset = 'utf8mb4'
)
cur = conn.cursor()
def city_pop_ratio(city_name):
    cur.execute(f""" SELECT c.Population, t.Population
                    FROM city c JOIN country t ON c.CountryCode = t.Code
                    WHERE c.Name='{city_name}'""")
    city_pop, country_pop = cur.fetchone() 
    return round(city_pop/country_pop*100,2)
```
